```python
import pandas as pd
import os

data_path = "../data1/raw/retail_clean_dataset"
sales_df = pd.read_csv(
    os.path.join(data_path, "sales_transactions.csv")
)

sales_df.shape
```

```python
sales_df.info()
```

```python
# Convert date column to datetime
sales_df['date'] = pd.to_datetime(sales_df['date'], errors='coerce')

# Check date range
print("Start date:", sales_df['date'].min())
print("End date:", sales_df['date'].max())

# Check invalid dates
print("Invalid dates:", sales_df['date'].isna().sum())
```

```python
# Check number of records for each year
sales_df['year'] = sales_df['date'].dt.year

print(sales_df['year'].value_counts().sort_index())
```

```python
print("Unique stores:", sales_df['store_id'].nunique())
print("Unique products:", sales_df['sku_id'].nunique())
print("Unique customers:", sales_df['customer_id'].nunique())
print("Unique receipts:", sales_df['receipt_id'].nunique())
print("Channels:", sales_df['channel'].unique())
```

```python
sales_df[['quantity', 'unit_price', 'total_value', 'discount_pct']].describe()
```

```python
print("Duplicate rows:", sales_df.duplicated().sum())
```

```python
print("Unique receipts:", sales_df['receipt_id'].nunique())
print("Total rows:", len(sales_df))
```

```python
print(sales_df['channel'].value_counts())
```

```python
print("Rows with promotion:", sales_df['promo_id'].notna().sum())
print("Rows without promotion:", sales_df['promo_id'].isna().sum())
```

```python
sales_df = sales_df.drop_duplicates()

print("Rows after removing duplicates:", len(sales_df))
print("Duplicates remaining:", sales_df.duplicated().sum())
```

```python
print("Quantity <= 0:", (sales_df['quantity'] <= 0).sum())
print("Unit price <= 0:", (sales_df['unit_price'] <= 0).sum())
print("Total value <= 0:", (sales_df['total_value'] <= 0).sum())
print("Discount < 0:", (sales_df['discount_pct'] < 0).sum())
print("Discount > 100:", (sales_df['discount_pct'] > 100).sum())
```

```python
sales_df.isna().sum()
```

```python
sales_df['promo_id'] = sales_df['promo_id'].fillna('No Promotion')
```

```python
print(sales_df['promo_id'].isna().sum())
print(sales_df['promo_id'].value_counts().head(10))
```

```python
print("Unique stores:", sales_df['store_id'].nunique())
print("Unique products:", sales_df['sku_id'].nunique())
print("Unique customers:", sales_df['customer_id'].nunique())

print("\nChannels:")
print(sales_df['channel'].value_counts())

print("\nMissing values:")
print(sales_df.isna().sum().sum())
```
